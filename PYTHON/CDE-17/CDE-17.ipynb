## CSV FILE HANDLING

```python
import csv
```

```python
competition = []
with open('competition.csv', 'r') as f:
    content = csv.reader(f)
    # print(content)
    for line in content:
        competition.append(line)

    print(competition)
```

```python
years = []
event = []
winner = []
with open('competition.csv', 'r') as f:
    content = csv.reader(f)
    # print(content)
    for line in competition[1:]:
        years.append(line[0])
        event.append(line[1])
        winner.append(line[2])
    print(years, event, winner)
```

```python
with open('competition.csv', 'a', newline ='') as f:
    writer_handler = csv.writer(f, delimiter = ',')
    writer_handler.writerow(['2012','Asia Cup', 'Pakistan'])
```

```python
with open('competition.csv', 'a', newline ='') as f:
    writer_handler = csv.writer(f, delimiter = ',')
    writer_handler.writerows([['2025','Asia Cup', 'Pakistan'],['2007','T20 WorldCup','India']])
```
