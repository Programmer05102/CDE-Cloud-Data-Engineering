## Object Oriented Programming (OOP)

### ABSTRACTION

```python
#Abstarct Class
class shape():
    #Abstract Method
    def sides(self):
        pass
```

```python
class square(shape):
    def sides(self):
        print(f"Square has 4 Sides.")
```

```python
sq_obj = square()
sq_obj.sides()
```

```python
class circle():
    def sides(self):
        print(f"Circle has No Sides.")
```

```python
crc_obj = circle()
crc_obj.sides()
```

### POLYMORPHISM
- The Ability of Different Classes to respond to the same message(method call) in different ways
- This Allows for code to work with objects of various classes as if they were objects of a common supeclass

```python
class Animal:
    def speak(self):
        return "Animal Speaks"
class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

        
#Create the Instances
anm = Animal()
dog = Dog()
cat = Cat()

#Speak Method With Different Objects
print(f"{anm.speak()}")
print(f"{dog.speak()}")
print(f"{cat.speak()}")
```

```python
class saylani:
    def __init__(self):
        pass
    def helpline(self):
        return '021'
    def _batch(self):
        return 'Python'
    def __result(self):
        return '85%'
```

#### Task 1: Create Three public,protected and private methods like we did for Attribute

```python
sa_obj = saylani()
print(sa_obj.helpline())
print(sa_obj._batch())
print(sa_obj._saylani__result())
```

#### TASK 2:
- use encapsulation for class i:e., StudentLogin():
- initialize two static but private attributes i:e., __user_name, __password
- create one private methods i:e., name __dbconectivity() that will validate you username & password. if both are valid then should print "valid user" otherwise "not valid"
- create one public methods i:e., update password & it should update your private password

```python
class StudentLogin:
    def __init__(self):
        self.__user_name = 'Hisham'
        self.__password = '1234'

    def __dbconnectivity(self, u, p):
        
        if u == self.__user_name and p == self.__password:
            print("Valid User")
            update_password(self, p)
        else:
            print("Not Valid User")

    def login(self, u, p):
        self.__dbconnectivity(u, p)

    def update_password(self, pa):
        pa = input("Enter Your New Password:")
        self.__password = pa
        print("Password Updated succesfully")
```

```python
std_login = StudentLogin()
u = input("Enter User Name: ")
p = input("Enter Your Password: ")
std_login.login(u, p)
```

## Data Files

- Read --> r
- Write --> w
- Append --> a
- Read/Write --> w+

```python
#METHOD-1 - We have to close File specifically
f = open('file.txt', 'w')
f.write('This is the content of my file')
f.close()
```

```python
# Method-2 = File will close automatically - Recommended
with open('file.txt', 'w') as f:
    f.write('Now it wil overwrite the previous text')
```

```python
with open('file.txt', 'a') as f:
    f.write('\nNow we are usibg append it wil not overwrite')
```

```python
with open('file.txt', 'a') as f:
    f.writelines('Now we are usibg append it wil not overwrite')
```

```python
with open('file.txt', 'r') as f:
    print(f.read())
```

### Task: Use mode = 'w+' and write something in the file first write and the read the updated content of the file

```python
with open('file1.txt', 'w+') as f:
    f.write('Now we are using append it wil not overwrite')
    f.seek(0) #f.seek(0) is imp otherwise it will read from the end of the file where write ends
    print(f.read())
```
