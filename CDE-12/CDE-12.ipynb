## Function within Function (Recursions)

```python
def fact(n):
    if n == 0 or n == 1: return 1
    else: return n * fact(n-1)
    
num = int(input("Enter Factorial Of: "))

print(fact(num))
```

## While Lopps

#### Task: Research About While Loop Syntax and Write a code of Finding Factorial Using While Loop

```python
# Task With Function
def fact(n):
    while n == 0 or n == 1: return 1
    else: return n * fact(n-1)

num = int(input("Enter Factorial Of: "))
print(fact(num))
```

```python
# Task Without Function
fact = 1
n = int(input("Enter Factorial Of: "))
while n > 0:
    fact *= n
    n -= 1

print(fact)
```

## Classes
- Everything in Python belongs some specific class
- Every class can have its **object**
- Every class have its own **methods**
- Every class have two parts:
  1. Characterstics/Properties -- > Attributes
  2. Behaviours/Actions --> Methods

```python
class cars():
    def __init__(self, color, model, make_year): # Constructor
        self.color = color
        self.model = model
        self.make_year = make_year
    
    def start(self): # Method
        """
        start object belongs to class car
        arg: no retun
        return: no return
        """
        print("Car Can Start")
    
    def stop(self): # Method
        """
        stop object belongs to class car
        arg: no retun
        return: no return
        """
        print("Car Can Stop")
```

```python
civic = cars('Black', 'Honda', 2025)
```

```python
civic.start()
```

```python
civic.stop()
```

```python
civic.color, civic.model, civic.make_year
```

####  Task: Create a Class which have Atleast 2 Attributes and Methods

```python
class human():
    def __init__(self, skin_color, disability):
        self.skin_color = skin_color
        self.disability = disability
    def walk(self):
        print("He Can Walk")

    def talk(self):
        print("He Can Talk")

hassan = human('Mild Fair','None')
hassan.walk()
hassan.talk()
```
