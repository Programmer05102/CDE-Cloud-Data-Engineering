```python
customers = [
 {
 "customer id": 0,
 "first name":"John",
 "last name": "Ogden",
 "address": "301 Arbor Rd.",
 },
 {
 "customer id": 1,
 "first name":"Ann",
 "last name": "Sattermyer",
 "address": "PO Box 1145",
 },
 {
 "customer id": 2,
 "first name":"Jill",
 "last name": "Somers",
 "address": "3 Main St.",
 },
 ]
```

```python
# Task-1 Method-1
for i in range(len(customers)):
    customers[i]['age'] = None
```

```python
# Task-1 Method-2
customers = [{**i, 'age':None} for i in customers]
```

```python
# Task-2 Method-1
lst = []
for i in customers:
    lst.append(list(i.values()))
```
