```python
class Parents:
    
    
    def __init__(self, eye_color, hair_color):
        self.eye_color = eye_color
        self.hair_color = hair_color

    def can_paint(self):
        print(f"Have Ability To Draw/Paint Arts")
```

```python
par_obj = Parents("Brown", "Brown")

print(par_obj.eye_color)
print(par_obj.hair_color)
par_obj.can_paint()
```

```python
class Child(Parents):
    pass
```

```python
child_obj = Child("Brown", "Black")

print(child_obj.eye_color)
print(child_obj.hair_color)
child_obj.can_paint()
```

```python
class Child(Parents):
    
    def __init__(self, eye_color, hair_color, skin_color):
        super().__init__(eye_color, hair_color)
        self.skin_color = skin_color

    def can_drive(self):
        print(f"Child Can Drive")
```

```python
child_obj = Child("Brown", "Black", "Brown")

print(child_obj.eye_color)
print(child_obj.hair_color)
print(child_obj.skin_color)
child_obj.can_paint()
child_obj.can_drive()
```

#### Task:
- Create a class with name "Father" which have two attributes and one method
- Create a class with name "Child" which inherit characterstics of "Father" but should have atleast two more attributes and methods other than "Father"

```python
class Father:

    def __init__(self, height, health):
        self.height = height
        self.health = health

    def can_swim(self):
        print(f"He can Swim")

class Child(Father):

    def __init__(self, height, health, iq, wealth):
        super().__init__(height, health)
        self.iq = iq
        self.wealth = wealth

    def can_drive(self):
        print(f"He can Drive")

    def can_shoot(self):
        print(f"He can Shot Great With Weapon")
        
```

```python
father_obj = Father(6, "Good")
print(father_obj.height)
print(father_obj.health)
father_obj.can_swim()

child_obj = Child(6, "Excellent", 100, "RichieRich")
print(child_obj.height)
print(child_obj.health)
child_obj.can_swim()
child_obj.can_drive()
child_obj.can_shoot
```

```python
#Multiple Class Inheritence

class Father:

    def __init__(self, height, health):
        self.height = height
        self.health = health

    def can_swim(self):
        print(f"He can Swim")

class Mother:
    
    def __init__(self, beauty, cooking):
        self.beauty = beauty
        self.cooking = cooking

    def can_sing(self):
        print(f"she can Sing")

class Child(Father, Mother):

    def __init__(self, height, health, beauty, cooking, iq, wealth):
        Father.__init__(self, height, health)
        Mother.__init__(self, beauty, cooking)
        self.iq = iq
        self.wealth = wealth

    def can_drive(self):
        print(f"He can Drive")

    def can_shoot(self):
        print(f"He can Shot Great With Weapon")
        
```

```python
child_obj = Child(6, "Excellent", "Handsome", "Skilled Cook", 100, "RichieRich")
print(child_obj.height)
print(child_obj.health)
print(child_obj.beauty)
print(child_obj.cooking)
print(child_obj.iq)
print(child_obj.wealth)
child_obj.can_swim()
child_obj.can_drive()
child_obj.can_shoot()
child_obj.can_sing()
```

## Overriding

```python
class Animal():
    def eating(self):
        print("Animals Can Eat")

class Birds(Animal):
    def eating(self):
        print("Birds Can Eat")
```

```python
ani_obj = Animal()
ani_obj.eating()
```

```python
birds_obj = Birds()
birds_obj.eating()
```

## Overloading

```python
from typing import overload

class Adder():


    # Abstraction
    @overload
    def add(self, x:int, y:int) -> int:
        ...

    @overload
    def add(self, x:float, y:float) -> float:
        ...

    @overload
    def add(self, x:str, y:str) -> str:
        ...


    # Actual Functionality
    def add(self, x, y):
        if isinstance(x, int) and isinstance(y, int):
            return x + y
        elif isinstance(x, float) and isinstance(y, float):
            return x + y
        elif isinstance(x, str) and isinstance(y, str):
            return x + y
        else:
            print("Type Error")
```

```python
adder_obj = Adder()
adder_obj.add(1, "2")
```
