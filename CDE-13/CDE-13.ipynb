## Lambda Functions 
- Without Name
- Never Been Called Before Or After
- No Return Keyword

```python
(lambda x, y: x + y)(8, 9)
```

```python
f = lambda x, y: x * y
```

```python
f(8, 5)
```

```python
f = lambda n: 1 if n == 1 else n * f(n-1)
```

```python
f(4)
```

```python
list(map(lambda x: (x, len(x)), ['Ali', 'Usama', 'Hassan', 'Qasim']))
```

```python
dict(filter(lambda x: x[1]>=50, [('Usama',50), ('Ali', 48), ("Hassan", 90)]))
```

## Classes

```python
class saylani():
    helpline = '0900808000' #static
    obj_count = 0 #dynamic
    
    def __init__(self, campus, staff_count, courses_count): #constructor
        self.campus = campus #attribute
        self.staff_count = staff_count
        self.courses_count = courses_count
        saylani.obj_count +=1 
        
    def details(self, courses): #method
        print(f"{self.campus} have {self.staff_count} and courses are {self.courses_count}")
        print("Courses: ",*courses)

    @staticmethod
    def facilities():
        print("Medical Also")
```

```python
za = saylani("ZA IT Park", 10, 5)
```

```python
za.details(["AI", "DE", "WEB"])
```

```python
za.obj_count
```

```python
za.helpline
```

```python
bahudurabad = saylani("Head Office", 11, 6)
```

```python
bahudurabad.obj_count
```

```python
#static method
saylani.facilities()
```

#### Task
Create a class of cars with atleast 3 attributes and 2 methods such that 1 method should be static. Also  class should have 2 class variables and one should dynamic

```python
class cars():
    helpline = '021-XXXXXXX'
    cars_count = 0

    def __init__(self, model, color, make):
        self.model = model
        self.make = make
        self.color = color
        cars.cars_count += 1

    def ignition(self):
        print(f"{self.color} {self.model} is Started")

    @staticmethod
    def movement():
        print("Brake is Released and Car Starts Moving")
```

```python
honda = cars("Civic", "Black", 2024)
honda.ignition()
cars.movement()
honda.cars_count
```

```python
suzuki = cars("Every", "White", 2010)
suzuki.ignition()
cars.movement()
suzuki.cars_count
```
